```python
79.
P0 = np.random.uniform(0,1,(5,2))
P1 = np.random.uniform(0,1,(5,2))
P  = np.random.uniform(0,1,(5,2))

# Vectorized: shape (lines, points)
d = P1 - P0  # (L,2)
# Cross products for all pairs
dp = P0[:,None,:] - P[None,:,:]  # (L,P,2)
cross = d[:,0,None]*dp[:,:,1] - d[:,1,None]*dp[:,:,0]
norm = np.sqrt((d**2).sum(axis=1))[:,None]
dist = np.abs(cross) / norm
print(dist.shape)  # (5,5)
```

```python
80.
def extract(Z, shape, position, fill=0):
    R = np.ones(shape, dtype=Z.dtype) * fill
    P = np.array(list(position))
    S = np.array(list(shape))
    Zshape = np.array(Z.shape)
    Rs = np.maximum(0, S//2 - P)
    Re = np.minimum(S, S//2 + Zshape - P)
    Zs = np.maximum(0, P - S//2)
    Ze = np.minimum(Zshape, P + S - S//2)
    R[Rs[0]:Re[0], Rs[1]:Re[1]] = Z[Zs[0]:Ze[0], Zs[1]:Ze[1]]
    return R

Z = np.arange(100).reshape(10,10)
print(extract(Z, shape=(5,5), position=(0,0), fill=0))
```

```python
88.
def game_of_life_step(Z):
    N = (np.roll(Z,1,0)+np.roll(Z,-1,0)+
         np.roll(Z,1,1)+np.roll(Z,-1,1)+
         np.roll(np.roll(Z,1,0),1,1)+np.roll(np.roll(Z,1,0),-1,1)+
         np.roll(np.roll(Z,-1,0),1,1)+np.roll(np.roll(Z,-1,0),-1,1))
    return (N==3) | (Z & (N==2))

Z = np.random.randint(0,2,(5,5))
print(game_of_life_step(Z))
```
